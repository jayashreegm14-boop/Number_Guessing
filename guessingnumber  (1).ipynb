```python
# A Number Guessing Name
Secret_Number = int(input("Enter Your Secret Number:"))
Guess = 0
Attempt = 0
while Guess != Secret_Number:
    Guess = int(input("Enter Your guessing Number:"))
    Attempt += 1
    if Guess > Secret_Number:
        print("Your Guessing is Too High")
    elif Guess < Secret_Number:
        print("You Guessing is Too Low")
    else:
        print("Congratulations ! You Guessed it! ")
print(f"\n You Guessed The Secret Number in {Attempt} moves.")
```
